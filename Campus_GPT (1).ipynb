**What is CampusGPT?**

CampusGPT is a Python-based AI chatbot project built using Hugging Face Transformers and basic Python functions. The project demonstrates how to:

Build a chatbot using a pre-trained language model
Generate AI-based responses
Store conversation history
Add chatbot personalities
Clean and format responses
Save chat logs
Analyze chatbot data using Pandas

This project is mainly designed for learning Generative AI and chatbot development.

```python
# Importing required libraries for building CampusGPT

from transformers import pipeline
import random
import textwrap
import gradio as gr
import pandas as pd
```

```python
chatbot = pipeline(
    "text-generation",   # Task type: tells the pipeline that we want to generate text based on input
    model="gpt2"         # Model name: GPT-2, a pre-trained Large Language Model for generating human-like text
)
print("Model loaded successfully")  # Confirms that the model is ready to use
```

```python
# Define a function that takes user input and returns a chatbot response
def basic_chat(user_input):

    response = chatbot(user_input, max_length=100)

    generated_text = response[0]['generated_text']

    # Return the generated text as the chatbot's reply
    return generated_text
```

```python

# This will keep track of everything the user and bot say
conversation_history = ""

# Define a function that takes user input and returns a response with memory
def chat_with_memory(user_input):
    global conversation_history
    conversation_history += f"User: {user_input}\n"
    response = chatbot(conversation_history, max_length=200)

    answer = response[0]['generated_text']
    conversation_history += f"Bot: {answer}\n"

    return answer
```

```python
def clean_response(text):
      latest_part = text.split("User:")[-1]
      cleaned_text = latest_part.strip()
      return cleaned_text
```

```python
def smart_brain(user_input):

    user_input = user_input.lower()

    # Greetings
    if "hello" in user_input or "hi" in user_input:
        return "Hello Student! Welcome to CampusGPT 😊"

    # Fees
    elif "fees" in user_input:
        return "MBA fees range between 5L and 12L depending on the college."

    # Placement
    elif "placement" in user_input:
        return "Average placement package ranges between 6-10 LPA."

    # Courses
    elif "courses" in user_input:
        return "Courses include Marketing, Finance, HR, Operations, and Business Analytics."

    # MBA
    elif "mba" in user_input:
        return "MBA is a professional degree focused on business and management skills."

    # Business Analytics
    elif "business analytics" in user_input:
        return "Business Analytics involves data analysis, visualization, and business decision-making."

    # AI
    elif "artificial intelligence" in user_input or "ai" in user_input:
        return "Artificial Intelligence enables machines to simulate human intelligence."

    # Machine Learning
    elif "machine learning" in user_input:
        return "Machine Learning is a subset of AI where systems learn from data."

    # Python
    elif "python" in user_input:
        return "Python is widely used in AI, Data Science, Automation, and Web Development."

    # SQL
    elif "sql" in user_input:
        return "SQL is used to manage and query databases."

    # Power BI
    elif "power bi" in user_input:
        return "Power BI is a business intelligence tool used for dashboards and data visualization."

    # Excel
    elif "excel" in user_input:
        return "Excel is used for data analysis, reporting, formulas, and dashboards."

    # Internship
    elif "internship" in user_input:
        return "Internships help students gain practical industry experience."

    # Resume
    elif "resume" in user_input:
        return "A good resume should include skills, education, projects, internships, and achievements."

    # Interview
    elif "interview" in user_input:
        return "Interview preparation should include technical skills, projects, and communication practice."

    # College
    elif "college" in user_input:
        return "Colleges provide academic learning, networking, and career opportunities."

    # Hostel
    elif "hostel" in user_input:
        return "Hostel facilities usually include rooms, mess, WiFi, and security."

    # Faculty
    elif "faculty" in user_input:
        return "Faculty members guide students academically and professionally."

    # Admission
    elif "admission" in user_input:
        return "MBA admissions generally require entrance exams, interviews, and academic records."

    # Scholarship
    elif "scholarship" in user_input:
        return "Scholarships are provided based on merit, category, or financial background."

    # Data Science
    elif "data science" in user_input:
        return "Data Science combines statistics, programming, and machine learning to extract insights from data."

    # Chatbot
    elif "chatbot" in user_input:
        return "A chatbot is an AI application that interacts with users through conversations."

    # Default AI Response
    else:
        return clean_response(chat_with_memory(user_input))
```

Personality Modes

```python

def personality_engine(user_input, mode):
    if mode == "mentor":
        return "Think strategically: " + smart_brain(user_input)
    elif mode == "professor":
        return "Answer academically: " + smart_brain(user_input)

    else:
        return smart_brain(user_input)
```

```python

chat_log = []

def log_chat(user, bot):

    chat_log.append({"user": user, "bot": bot})

    return pd.DataFrame(chat_log)
```

```python
def full_chat(user_input, mode):

    # It generates the chatbot response based on user input and selected mode
    response = personality_engine(user_input, mode)

    # Save the conversation (user input + bot response) into chat log
    log_chat(user_input, response)

    # Return the final response to display in the UI
    return response

iface = gr.Interface(

    # fn = function that will run when user submits input
    fn=full_chat,

    # inputs = list of input components for the UI
    inputs=[
        gr.Textbox(label="Enter your query"),
        # Textbox allows user to type their question

        gr.Dropdown(["normal", "mentor", "professor"], label="Mode")
        # Dropdown allows user to select chatbot personality
    ],

    # outputs = what the UI will display after processing
    outputs="text",  # Output will be shown as plain text

    title="CampusGPT 3.0",

    description="AI Assistant for College Queries"
)
iface.launch()
```
